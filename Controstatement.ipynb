```python
#contsional statement
mark=int(input("enter the mark"))
if(mark>=90 and mark <=100):
    print("pass")
    print("A")
elif(mark>=80 and mark <=89):
    print("Pass")
    print("B")
elif(mark>=70 and mark <=79):
    print("Pass")
    print("C")
elif(mark>=60 and mark <=69):
    print("Pass")
    print("D")
elif(mark>=41 and mark <=59):
    print("Pass")
    print("E")
elif(mark>=0 and mark <=41):
    print("Fail")
else:
    print("absent")
```
