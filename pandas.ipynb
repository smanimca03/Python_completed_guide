```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("employee_salary_dataset copy.csv")
plt.figure(figsize=(10,5))
plt.plot(df["Experience_Years"],df["Monthly_Salary"],color="orange",marker="s",linestyle="-.",linewidth=1)
plt.title("Salary Analysis")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("employee_salary_dataset copy.csv")
plt.figure(figsize=(10,5))
plt.bar(df["Experience_Years"],df["Monthly_Salary"],color="orange",linestyle="--",linewidth=1)
plt.title("Salary Analysis")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```
