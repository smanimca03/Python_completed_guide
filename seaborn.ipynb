```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("employee_salary_dataset.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
sns.scatterplot(x="Age",y="Monthly_Salary",data=df)
plt.show()
```

```python
sns.lineplot(x="Age",y="Monthly_Salary",data=df)
plt.show()
```

```python
sns.barplot(x="Age",y="Monthly_Salary",data=df)
plt.show()
```
